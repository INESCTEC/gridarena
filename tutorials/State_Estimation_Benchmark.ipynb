# State Estimation Benchmark – Tutorial Notebook

This notebook walks you through how to interact with the **State Estimation API** end-to-end:

1. Create a small **mock grid** and corresponding **historical data** and insert them via the API.
2. Use the **State Estimation endpoints** to:
   - Download training data full voltage measurements
   - Download test data without voltage measurements in randomly selected nodes in a single timestep
   - Build and submit the **guess** for the voltage measurements in the provided timestep
   - Retrieve a **score** for your guess

The goal is to provide a minimal, reproducible example that you can adapt to your own grids, models and workflows.

## 0. Environment and prerequisites

Before running this notebook, you should have:

- A running instance of the **State Estimation API** (locally or remote).
- The correct `BASE_URL` and path prefixes configured in the code cells below.
- Network access from this notebook environment to the API endpoint.

The helper functions defined in the next cells (e.g. `check_response`, `safe_get_json`) are thin wrappers around `requests` that:
- Simplify HTTP calls
- Standardise error handling
- Return already parsed JSON where appropriate

You do **not** need to modify these helpers for normal use; they are here to keep the rest of the tutorial cleaner.

## Auxiliary Functions and Variable Declaration

```python
import requests
import pandas as pd
from collections import defaultdict
from pathlib import Path
import json
import numpy as np
import topologytest as tp
import matplotlib.pyplot as plt
import random
from datetime import datetime

BASE_URL = "http://localhost:8080"           # adapt if API initialized in a different port
STATE_ESTIMATION_PREFIX = "/state_estimation_benchmark"


def check_response(resp: requests.Response):
    """Raise for HTTP errors and return parsed JSON or raw text."""
    try:
        resp.raise_for_status()
    except requests.HTTPError as e:
        try:
            print("Error payload:", resp.json())
        except Exception:
            print("Raw response:", resp.text)
        raise e
    try:
        return resp.json()
    except Exception:
        return resp.text
```

## 0.3 Helper functions to insert data via the API

In the next cells, we define two helper functions:

- `register_grid_from_file(json_path)` – reads a grid JSON file and calls the `/grid` endpoint.
- `register_historical_from_file(json_path)` – reads a historical JSON file and calls the `/historical` endpoint.

These functions:

- Encapsulate the HTTP logic (HTTP method, headers and payload).
- Log basic information about the request / response.
- Raise an error if the API does not accept the data.

If you already have your grid and historical data in separate JSON files, you can reuse these helpers without changes.

```python
def register_grid_from_file(json_path: str | Path):
    """
    Upload a GridData JSON file to POST /grid/.
    """
    
    json_path = Path(json_path)
    if not json_path.exists():
        raise FileNotFoundError(json_path)
    url = f"{BASE_URL}/grid/"
    with open(json_path, "rb") as f:
        files = {"file": (json_path.name, f, "application/json")}
        resp = requests.post(url, files=files)
    return check_response(resp)


def register_historical_from_file(json_path: str | Path):
    """
    Upload a HistoricalData JSON file to POST /historical/.
    """
    
    json_path = Path(json_path)
    if not json_path.exists():
        raise FileNotFoundError(json_path)
    url = f"{BASE_URL}/historical/"
    with open(json_path, "rb") as f:
        files = {"file": (json_path.name, f, "application/json")}
        resp = requests.post(url, files=files)
    return check_response(resp)
```

# 1. State Estimation: API Usage

In this section, we demonstrate the **typical workflow** for the state estimation benchmark:

1. Download training data (with known voltage measurements) for a given grid with configurable noise and observability settings.
2. Download a masked **state estimation input** with configurable noise and observability settings.
3. Build a simple voltage estimate payload for the unknown nodes.
4. Submit the estimates to the API.
5. Request the resulting score.

We will start with a very simple (and deliberately bad) random baseline, just to validate the request / response flow.
You can later replace it with a real estimation model.

## 1.1 Download training data

**Endpoint:**  
`GET {STATE_ESTIMATION_PREFIX}/training-state-data?noise_difficulty=clean&observability=medium`

This endpoint returns:

- The **nodes** in the grid for which we have historical data.
- For each node, a **time-series** of measurements.
- Metadata that may include which grids are public or private.

Important details:
- noise_difficulty controls the measurement noise profile
- observability is accepted by the endpoint, even though the returned training payload is still structured as historical masked measurements

Usage in this tutorial:

- We call it **once**, for `grid_001`.
- We store the result as `training_data`.

```python
def fetch_training_se_data(
    noise_difficulty: str = "clean",
    observability: str = "medium",
) -> dict:
    """
    Call GET /training-state-data and return the nested JSON as Python dict.
    """
    url = f"{BASE_URL}{STATE_ESTIMATION_PREFIX}/training-state-data"
    resp = requests.get(
        url,
        params={
            "noise_difficulty": noise_difficulty,
            "observability": observability,
        },
    )
    return check_response(resp)
```

```python
training_data = fetch_training_se_data(
    noise_difficulty="clean",
    observability="medium",
)
```

## 1.3 Building an estimate payload

To submit a state estimation guess to the benchmark, we must send a JSON object with the following structure:

- `user_id` – your benchmark user identifier
- `estimation_id` – An arbitrary identifier chosen by you (e.g. `"tutorial_guess"`).  
- `grid_id` – The grid to which this guess applies (here, `"grid_001"`).
- `timestamp` – The timestamp to which we are trying to guess the voltages.  
- `estimates` – A list of voltage measurement guesses `{single_guess}`, with:
  - `masked_node_id` - identifier of the node where we are guessing the voltage.
  - `voltage_magnitude` - predicted voltage magnitude.
  - `voltage_angle` - predicted voltage angle.

Important:
- the target nodes are the masked node IDs listed under:
  - `test_data["Current State"]["unknown_nodes"]`
- the known nodes are already provided in:
  - `test_data["Current State"]["known_nodes"]`
- the measurement history is provided in:
  - `test_data["Measurement Data"]`

In this notebook we:

- Extract the list of nodes from the test dataset.
- Assign each node a random voltage measurement.
- Store the result in the `guess` dictionary, ready to be sent to the API.

```python
USER_ID = "test_user"  # put your benchmark user_id here

# 1) Use the estimation task metadata returned by the API
estimation_id = test_data["Estimation Task ID"]
masked_grid_id = test_data["Masked GridID"]

# 2) Build estimates only for unknown nodes
unknown_nodes = test_data["Current State"]["unknown_nodes"]

# 3) Pick a timestamp for the submission
#    We use the latest timestamp available in the historical measurement data.
all_timestamps = []
for measurements in test_data["Measurement Data"].values():
    for m in measurements:
        all_timestamps.append(datetime.fromisoformat(m["datetime"]))

if all_timestamps:
    target_timestamp = max(all_timestamps)
else:
    target_timestamp = datetime.utcnow()

# 4) Build a simple random baseline
random.seed(42)

estimates = []
for node_id in sorted(unknown_nodes):
    estimates.append(
        {
            "masked_node_id": node_id,
            "voltage_magnitude": random.uniform(220.0, 240.0),
            "voltage_angle": random.uniform(-10.0, 10.0),
        }
    )

# 5) Assemble final payload
guess = {
    "user_id": USER_ID,
    "estimation_id": estimation_id,
    "grid_id": masked_grid_id,
    "timestamp": target_timestamp.isoformat(timespec="seconds"),
    "estimates": estimates,
}
```

## 1.4 Submitting the estimates to the API

Once we have our `guess` object, we need to send it to the **submission endpoint**:

**Endpoint:**  
`POST {STATE_ESTIMATION_PREFIX}/submit-estimates`

which:

- Validates the format of the payload.
- Registers the guess under the provided `estimation_id`.
- Returns a confirmation status.

If the submission is accepted, we can later **request a score** for this `user_id` and `grid_id`.  
If it fails, the error message returned here is usually the best place to troubleshoot the JSON structure.

```python
def submit_se_data_guesses(guess: dict) -> dict:
    """
    Call POST /submit-estimates and return the parsed response.
    """
    url = f"{BASE_URL}{STATE_ESTIMATION_PREFIX}/submit-estimates"
    resp = requests.post(url, json=guess)
    return check_response(resp)
```

```python
submit_se_data_guesses(guess)
```

## 1.5 Requesting a score for the estimate

**Endpoint:**  
`GET {STATE_ESTIMATION_PREFIX}/state-score?user_id=...&estimation_id=...`

This endpoint:
- looks up the estimates previously submitted for the given `user_id` and `estimation_id`
- compares them with the hidden ground truth
- Returns one or more **metrics** (for example, accuracy per node or overall accuracy).

In this tutorial we:

- Request the score for the random `tutorial_guess`.
- Inspect the raw response and discuss how you might log or visualise these metrics when using a real model.

```python
def fetch_se_score(user_id: str, estimation_id: str) -> dict:
    """
    Fetch the state estimation score for a given user_id and estimation_id.
    """
    url = f"{BASE_URL}{STATE_ESTIMATION_PREFIX}/state-score"
    resp = requests.get(
        url,
        params={"user_id": user_id, "estimation_id": estimation_id},
    )
    return check_response(resp)
```

## 2. State Estimation: UI Usage

The web interface provides a point-and-click version of the benchmark workflow, mirroring the API calls shown earlier. You can use it to run the full State Estimation pipeline without writing any code.

1. **Open the State Estimation page**

   From the application UI, open **State Estimation**.

   On the page you will see three sections:
   - **Download Data**
   - **Submit State Estimates**
   - **Check Score**

2. **Download data**

   2. **Download data**

   In the **Download Data** section you can:
   - choose a **Noise difficulty**: `clean`, `easy`, `medium`, or `hard`
   - choose an **Observability** level: `clean`, `easy`, `medium`, or `hard`
   - click **Download Training Data**
   - enter a **User ID** and click **Download Estimation Input**

   These files are what you will feed into your own algorithm or model.

   Notes:
   - estimation input is generated for the provided `user_id`
   - `noise_difficulty` changes the measurement noise profile
   - `observability` changes how much of the current state and history is visible

3. **Submit state estimates**

   After your algorithm has produced voltage estimates for the unknown masked nodes, use the **Submit State Estimates** section:
   - enter your **User ID**
   - enter the **Estimation Task ID**
   - Use the file picker to select your guesses file and click **Submit Guesses**.  
   If the format is valid, the UI will confirm that the submission was accepted.

4. **Check score**

   Once a guess has been submitted, you can evaluate it in the **Check Score** section:
   - Enter the same **USER ID** and **Estimation Task ID** you used when submitting.
   - Click **Check Score** to request the evaluation.
   - The UI will display the resulting metric(s), which you can compare across different models or runs.

You can repeat the *submit → score* cycle as many times as you like, using different models or hyper-parameters, while keeping the same downloaded datasets. This makes it easy to treat the interface as a lightweight leaderboard for your State Estimation experiments.
