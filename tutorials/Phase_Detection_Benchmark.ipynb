# Phase Detection Benchmark – Tutorial Notebook

This notebook walks you through how to interact with the **Phase Detection API** end-to-end:

1. Create a small **mock grid** and corresponding **historical data** and insert them via the API.
2. Use the **phase identification endpoints** to:
   - Download training data with labelled phases
   - Download test data without phases
   - Build and submit a phase **guess** for each node
   - Retrieve a **score** for your guess

The goal is to provide a minimal, reproducible example that you can adapt to your own grids, models and workflows.

## 0. Environment and prerequisites

Before running this notebook, you should have:

- A running instance of the **Phase Detection API** (locally or remote).
- The correct `BASE_URL` and path prefixes configured in the code cells below.
- Network access from this notebook environment to the API endpoint.

The helper functions defined in the next cells (e.g. `check_response`, `safe_get_json`) are thin wrappers around `requests` that:
- Simplify HTTP calls
- Standardise error handling
- Return already parsed JSON where appropriate

You do **not** need to modify these helpers for normal use; they are here to keep the rest of the tutorial cleaner.

## Auxiliary Functions and Variable Declaration

````python
import requests
import pandas as pd
from collections import defaultdict
from pathlib import Path
import json
import numpy as np

BASE_URL = "http://localhost:8080"           # adapt if API initialized in a different port
PHASE_PREFIX = "/phase_identification_benchmark"


def check_response(resp: requests.Response):
    """Raise for HTTP errors and return parsed JSON or raw text."""
    try:
        resp.raise_for_status()
    except requests.HTTPError as e:
        try:
            print("Error payload:", resp.json())
        except Exception:
            print("Raw response:", resp.text)
        raise e
    try:
        return resp.json()
    except Exception:
        return resp.text
````

## 0.3 Helper functions to insert data via the API

In the next cells, we define two helper functions:

- `register_grid_from_file(json_path)` – reads a grid JSON file and calls the `/grid` endpoint.
- `register_historical_from_file(json_path)` – reads a historical JSON file and calls the `/historical` endpoint.

These functions:

- Encapsulate the HTTP logic (HTTP method, headers and payload).
- Log basic information about the request / response.
- Raise an error if the API does not accept the data.

If you already have your grid and historical data in separate JSON files, you can reuse these helpers without changes.

````python
def register_grid_from_file(json_path: str | Path):
    """
    Upload a GridData JSON file to POST /grid/.
    """
    
    json_path = Path(json_path)
    if not json_path.exists():
        raise FileNotFoundError(json_path)
    url = f"{BASE_URL}/grid/"
    with open(json_path, "rb") as f:
        files = {"file": (json_path.name, f, "application/json")}
        resp = requests.post(url, files=files)
    return check_response(resp)


def register_historical_from_file(json_path: str | Path):
    """
    Upload a HistoricalData JSON file to POST /historical/.
    """
    
    json_path = Path(json_path)
    if not json_path.exists():
        raise FileNotFoundError(json_path)
    url = f"{BASE_URL}/historical/"
    with open(json_path, "rb") as f:
        files = {"file": (json_path.name, f, "application/json")}
        resp = requests.post(url, files=files)
    return check_response(resp)
````

# 1. Phase Identification: API Usage

In this section, we demonstrate the current benchmark workflow exposed by the API:

1. Download **training data** with known phases using `/training-phase-data`.
2. Download **test data** with anonymised keys using `/phase-data`.
3. Build a simple phase guess for one grid from the anonymised test keys.
4. Submit that guess to the API.
5. Request the resulting score.

We will use a deliberately weak random baseline just to validate the request / response flow.
You can later replace it with a real model.

## 1.1 Download training data with phases

**Endpoint:**  
`GET {PHASE_PREFIX}/training-phase-data?difficulty=clean`

This endpoint returns:

- The **nodes** in the grid for which we have historical data.
- For each node, a **time-series** of measurements **with known phases**.
- Metadata that may include which grids are public or private.

Important details:

- the response is grouped by grid_id and then by node_id
- training records include the true phase
- difficulty controls the corruption profile applied to numeric values
    - Use `clean` to inspect the original benchmark structure
    - `easy` / `medium` / `hard` to evaluate how robust your method is to progressively noisier inputs.

Usage in this tutorial:

- We call it **once**, for `grid_001`.
- We store the result as `training_data`.
- We then explore the structure (keys, shapes, number of nodes) to understand how a model would consume it.

````python
def fetch_training_phase_data(difficulty: str = "clean") -> dict:
    """
    Call GET /training-phase-data and return the nested JSON as Python dict.
    """
    url = f"{BASE_URL}{PHASE_PREFIX}/training-phase-data"
    resp = requests.get(url, params={"difficulty": difficulty})
    return check_response(resp)
````

````python
training_data = fetch_training_phase_data()
````

````python
for grid_id, nodes in training_data["grids"].items():
    num_nodes = len(nodes)
    num_meas = sum(len(meas_list) for meas_list in nodes.values())
    print(f"Grid {grid_id}: {num_nodes} nodes, {num_meas} measurements")
````

````python
from datetime import datetime

for grid_id, nodes in training_data["grids"].items():
    print(f"\nGrid {grid_id}")
    for node_id, measurements in nodes.items():
        times = [datetime.fromisoformat(m["datetime"]) for m in measurements]
        t_min, t_max = min(times), max(times)
        print(f"  Node {node_id}: {len(measurements)} points "
              f"from {t_min} to {t_max}")
````

````python
def avg(xs):
    return sum(xs) / len(xs) if xs else None

for grid_id, nodes in training_data["grids"].items():
    print(f"\nGrid {grid_id}")
    for node_id, measurements in nodes.items():
        p = [m["power_active"] for m in measurements]
        v = [m["voltage_magnitude"] for m in measurements]

        print(
            f"  Node {node_id}: "
            f"P_active avg={avg(p):.1f}, min={min(p):.1f}, max={max(p):.1f}; "
            f"V_mag avg={avg(v):.1f}, min={min(v):.1f}, max={max(v):.1f}"
        )
````

````python
for grid_id, nodes in training_data["grids"].items():
    print(f"\nGrid {grid_id}")
    for node_id, measurements in nodes.items():
        phases = {m["phase"] for m in measurements}
        print(f"  Node {node_id}: phases present = {sorted(phases)}")
````

This endpoint follows a similar structure to the training data, but **without** the true phase labels. Each series is keyed by an **anonymised key**. This is the data on which you must **predict** the phase for each node.

`Difficulty` controls the measurement corruption profile applied by the platform. Use `clean` to inspect the original benchmark structure, and `easy` / `medium` / `hard` to evaluate how robust your method is to progressively noisier inputs.

## 1.3 Building a guess payload

To submit a phase assignment to the benchmark, we must send a JSON object with the following structure:

- `guess_id` – An arbitrary identifier chosen by you (e.g. `"tutorial_guess"`).  
- `grid_id` – The grid to which this guess applies (here, `"grid_001"`).  
- `guesses` – A mapping `{anonymised_key: phase}`, where:
  - `anonymised_key` represents one of the nodes present in the test data.
  - `phase` is one of `"R"`, `"S"` or `"T"`.

In this notebook we:

- Extract the list of nodes from the test dataset.
- Assign each node a random phase.
- Store the result in the `guess` dictionary, ready to be sent to the API.

Important: for the test dataset, you do **not** submit node IDs.
You submit guesses for the anonymised keys returned by the API.

````python
guess = {"guess_id" : "tutorial_guess", "grid_id" : "grid_001", "guesses": {}}

for anon_key in test_data["grids"]["grid_001"]:
    guess["guesses"][anon_key] = np.random.choice(['R', 'S', 'T'])
````

## 1.4 Submitting the guess to the API

Once we have our `guess` object, we need to send it to the **submission endpoint**:

**Endpoint:**  
`POST {PHASE_PREFIX}/submit-results`

 which:

- Validates the format of the payload.
- Registers the guess under the provided `guess_id`.
- Returns a confirmation status.

If the submission is accepted, we can later **request a score** for this `guess_id` and `grid_id`.  
If it fails, the error message returned here is usually the best place to troubleshoot the JSON structure.

````python
def submit_phase_data_guesses(guess: dict) -> dict:
    """
    Call POST /submit-results and return the parsed response.
    """
    url = f"{BASE_URL}{PHASE_PREFIX}/submit-results"
    resp = requests.post(url, json=guess)
    return check_response(resp)
````

````python
submit_phase_data_guesses(guess)
````

## 1.5 Requesting a score for the guess

**Endpoint:**  
`GET {PHASE_PREFIX}/score?guess_id=...&grid_id=...`

This endpoint:

- Looks up the stored guess identified by `guess_id` and `grid_id`.
- Compares it against the **hidden ground truth** for the test data.
- Returns one or more **metrics** (for example, accuracy per node or overall accuracy).

In this tutorial we:

- Request the score for the random `tutorial_guess`.
- Inspect the raw response and discuss how you might log or visualise these metrics when using a real model.

````python
def fetch_phase_score(guess_id: str,
                      grid_id: str):
    """
    Fetch the phase detection score for a given guess_id and grid_id.

    Relies on global BASE_URL and PHASE_PREFIX variables.
    Returns the raw `requests.Response` object.
    """
    url = f"{BASE_URL}{PHASE_PREFIX}/score"
    payload = {"guess_id": guess_id, "grid_id": grid_id}
    resp = requests.get(url, params=payload)
    return resp.text
````

````python
fetch_phase_score("tutorial_guess", "grid_001")
````

## 2. Phase Identification: UI Usage

The web interface provides a point-and-click version of the benchmark workflow, mirroring the API calls shown earlier. You can use it to run the full phase identification pipeline without writing any code.

1. **Open the Phase Identification page**

   From the application UI, open **Phase Identification**.

   On the page you will see three sections:
   - **Download Data**
   - **Submit Phase Guesses**
   - **Check Score**

2. **Download data**

   In the **Download Data** section you can:
   - choose a **Difficulty**: `clean`, `easy`, `medium`, or `hard`
   - click **Download Training Data**
   - click **Download Phase Data**

   Notes:
   - training data includes true phases
   - phase data uses anonymised keys and hides the true phases
   - downloads are returned as JSON files generated by the browser

   These files are what you will feed into your own algorithm or model.

3. **Submit phase guesses**

   In the **Submit Phase Guesses** section:
   - enter a **User ID**
   - enter a **Grid ID**
   - choose a JSON file containing your guesses
   - click **Submit Results**

   The uploaded JSON file should represent the `guesses` mapping itself, for example:

   ```json
   {
     "anon_key_001": "R",
     "anon_key_002": "T",
     "anon_key_003": "S"
   }

4. **Check score**

   Once a guess has been submitted, you can evaluate it in the **Check Score** section:
   - enter the same **User ID**
   - enter the same **Grid ID**
   - Click **Check Score** to request the evaluation.
   - The UI will display the resulting metric(s), which you can compare across different models or runs.

You can repeat the *submit → score* cycle as many times as you like, using different models or hyper-parameters, while keeping the same downloaded datasets. This makes it easy to treat the interface as a lightweight leaderboard for your phase identification experiments.
